```python
%load_ext autoreload
%autoreload 2

import sys
sys.path.append("..")

from src.CocoDataset import CocoDataset
from src.geometria import puntoApoyo
from src.viz import dibujarCajas, dibujarPuntos
import matplotlib.pyplot as plt

ds = CocoDataset("../data/raw/football-players/valid")

IMG_ID = [10,11,12,13,14]

for IMG in IMG_ID:
    cajas10 = ds.cajas(IMG)

    puntos = []

    for caja, clase in cajas10:
        if clase != "ball":
            puntos.append(puntoApoyo(caja))

    fig, ax = plt.subplots(figsize=(14, 14))
    ax.imshow(ds.imagen(IMG))
    ax.axis("off")
    dibujarCajas(ax, cajas10)
    dibujarPuntos(ax, puntos)
    plt.show()

    fig, ax = plt.subplots(figsize=(14, 14))
    ax.imshow(ds.imagen(IMG))
    ax.axis("off")
    dibujarPuntos(ax, puntos)
    plt.show()
```
